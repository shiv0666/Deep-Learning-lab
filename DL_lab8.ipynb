```python
# ============================================================
# STOCK MARKET PREDICTION USING LSTM
# Dataset:
# borismarjanovic/price-volume-data-for-all-us-stocks-etfs
#
# Activities:
# 1. Prepare sequential data
# 2. Implement LSTM network
# 3. Visualize predicted vs actual values
# ============================================================


# ============================================================
# 1. INSTALL KAGGLEHUB
# ============================================================

!pip install -q kagglehub


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import kagglehub

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 3. CHECK TENSORFLOW / GPU
# ============================================================

print("TensorFlow Version:", tf.__version__)

print("GPU Available:")
print(tf.config.list_physical_devices("GPU"))


# ============================================================
# 4. DOWNLOAD DATASET DIRECTLY FROM KAGGLE
# ============================================================

path = kagglehub.dataset_download(
    "borismarjanovic/price-volume-data-for-all-us-stocks-etfs"
)

print("\nDataset downloaded successfully!")

print("Dataset path:")
print(path)


# ============================================================
# 5. FIND STOCK FILES
# ============================================================

print("\nSearching for stock files...")

stock_files = []

for root, dirs, files in os.walk(path):

    for file in files:

        if file.endswith(".txt") or file.endswith(".csv"):

            stock_files.append(
                os.path.join(root, file)
            )


print("Number of files found:", len(stock_files))

print("\nFirst 10 files:")

for file in stock_files[:10]:

    print(file)


# ============================================================
# 6. FIND APPLE STOCK FILE
# ============================================================

apple_file = None

for file in stock_files:

    filename = os.path.basename(file).lower()

    if filename == "aapl.us.txt":

        apple_file = file

        break


print("\nApple stock file:")

print(apple_file)


# ============================================================
# 7. LOAD APPLE STOCK DATA
# ============================================================

if apple_file is None:

    raise FileNotFoundError(
        "AAPL.us.txt was not found in the dataset."
    )


df = pd.read_csv(
    apple_file
)


# ============================================================
# 8. DISPLAY DATA
# ============================================================

print("\nFirst 5 rows:")

print(df.head())


print("\nDataset Shape:")

print(df.shape)


print("\nColumns:")

print(df.columns)


# ============================================================
# 9. CONVERT DATE
# ============================================================

df["Date"] = pd.to_datetime(
    df["Date"]
)


# Sort according to date
df = df.sort_values(
    "Date"
)


# Reset index
df = df.reset_index(
    drop=True
)


# ============================================================
# 10. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")

print(
    df.isnull().sum()
)


# ============================================================
# 11. DISPLAY STOCK PRICE
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    df["Date"],
    df["Close"]
)

plt.xlabel("Date")

plt.ylabel("Closing Price")

plt.title(
    "Apple Stock Closing Price"
)

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 12. USE CLOSE PRICE FOR LSTM
# ============================================================

data = df[
    ["Close"]
].values


print(
    "\nTotal closing price records:",
    len(data)
)


# ============================================================
# 13. NORMALIZE DATA
# ============================================================

scaler = MinMaxScaler(
    feature_range=(0, 1)
)


scaled_data = scaler.fit_transform(
    data
)


print("\nData normalized successfully!")


# ============================================================
# 14. TRAIN / TEST SPLIT
# ============================================================

train_size = int(
    len(scaled_data) * 0.80
)


train_data = scaled_data[
    :train_size
]


test_data = scaled_data[
    train_size:
]


print(
    "\nTraining records:",
    len(train_data)
)

print(
    "Testing records:",
    len(test_data)
)


# ============================================================
# 15. CREATE SEQUENTIAL DATA
# ============================================================

# Number of previous days used
# to predict the next day

time_step = 60


def create_sequences(data, time_step):

    X = []

    y = []

    for i in range(
        time_step,
        len(data)
    ):

        X.append(
            data[
                i-time_step:i,
                0
            ]
        )

        y.append(
            data[
                i,
                0
            ]
        )

    return np.array(X), np.array(y)


# ============================================================
# 16. CREATE TRAINING SEQUENCES
# ============================================================

X_train, y_train = create_sequences(

    train_data,

    time_step

)


# ============================================================
# 17. CREATE TESTING SEQUENCES
# ============================================================

# Add previous 60 training values
# so that the first test sequence
# has enough historical information

test_input = np.concatenate(

    (
        train_data[-time_step:],
        test_data
    )

)


X_test, y_test = create_sequences(

    test_input,

    time_step

)


print("\n======================================")
print("SEQUENTIAL DATA")
print("======================================")

print(
    "X_train shape:",
    X_train.shape
)

print(
    "y_train shape:",
    y_train.shape
)

print(
    "X_test shape:",
    X_test.shape
)

print(
    "y_test shape:",
    y_test.shape
)


# ============================================================
# 18. RESHAPE DATA FOR LSTM
# ============================================================

# LSTM requires:
#
# Samples
# Time Steps
# Features

X_train = X_train.reshape(

    X_train.shape[0],

    X_train.shape[1],

    1

)


X_test = X_test.reshape(

    X_test.shape[0],

    X_test.shape[1],

    1

)


print("\nFinal LSTM input shape:")

print(
    X_train.shape
)


# ============================================================
# 19. BUILD LSTM MODEL
# ============================================================

model = Sequential()


# First LSTM layer
model.add(

    LSTM(

        64,

        return_sequences=True,

        input_shape=(

            time_step,

            1

        )

    )

)


# Dropout
model.add(

    Dropout(0.2)

)


# Second LSTM layer
model.add(

    LSTM(

        64,

        return_sequences=False

    )

)


# Dropout
model.add(

    Dropout(0.2)

)


# Output layer
model.add(

    Dense(

        1

    )

)


# ============================================================
# 20. DISPLAY MODEL
# ============================================================

model.summary()


# ============================================================
# 21. COMPILE MODEL
# ============================================================

model.compile(

    optimizer="adam",

    loss="mean_squared_error"

)


# ============================================================
# 22. EARLY STOPPING
# ============================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=5,

    restore_best_weights=True

)


# ============================================================
# 23. TRAIN LSTM
# ============================================================

print("\n======================================")
print("TRAINING LSTM MODEL")
print("======================================")


history = model.fit(

    X_train,

    y_train,

    validation_split=0.1,

    epochs=30,

    batch_size=32,

    callbacks=[
        early_stopping
    ],

    verbose=1

)


# ============================================================
# 24. PLOT TRAINING LOSS
# ============================================================

plt.figure(figsize=(8, 5))


plt.plot(

    history.history["loss"],

    label="Training Loss"

)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss"

)


plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "LSTM Training and Validation Loss"
)


plt.legend()

plt.show()


# ============================================================
# 25. MAKE PREDICTIONS
# ============================================================

predictions = model.predict(

    X_test

)


# ============================================================
# 26. CONVERT PREDICTIONS BACK
# TO ORIGINAL STOCK PRICE
# ============================================================

predicted_prices = scaler.inverse_transform(

    predictions

)


actual_prices = scaler.inverse_transform(

    y_test.reshape(
        -1,
        1
    )

)


# ============================================================
# 27. CALCULATE METRICS
# ============================================================

mae = mean_absolute_error(

    actual_prices,

    predicted_prices

)


rmse = np.sqrt(

    mean_squared_error(

        actual_prices,

        predicted_prices

    )

)


r2 = r2_score(

    actual_prices,

    predicted_prices

)


print("\n======================================")
print("MODEL PERFORMANCE")
print("======================================")


print(

    "MAE:",
    round(mae, 4)

)


print(

    "RMSE:",
    round(rmse, 4)

)


print(

    "R² Score:",
    round(r2, 4)

)


# ============================================================
# 28. VISUALIZE ACTUAL VS PREDICTED
# ============================================================

test_dates = df[
    "Date"
].iloc[
    train_size:
].values


plt.figure(figsize=(14, 6))


plt.plot(

    test_dates,

    actual_prices,

    label="Actual Price"

)


plt.plot(

    test_dates,

    predicted_prices,

    label="Predicted Price"

)


plt.xlabel("Date")

plt.ylabel("Stock Price")


plt.title(

    "Actual vs Predicted Apple Stock Price using LSTM"

)


plt.legend()

plt.xticks(rotation=45)

plt.show()


# ============================================================
# 29. DISPLAY LAST 20 PREDICTIONS
# ============================================================

results = pd.DataFrame({

    "Date":
        test_dates,

    "Actual Price":
        actual_prices.flatten(),

    "Predicted Price":
        predicted_prices.flatten()

})


print("\n======================================")
print("LAST 20 PREDICTIONS")
print("======================================")


print(
    results.tail(20)
)


# ============================================================
# 30. SAVE MODEL
# ============================================================

model.save(

    "/content/stock_lstm_model.h5"

)


print(

    "\nModel saved successfully!"

)

print(

    "Location: /content/stock_lstm_model.h5"

)
```
