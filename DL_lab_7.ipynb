```python
# ============================================================
# DOG BREED CLASSIFICATION USING TRANSFER LEARNING
# Dataset: hartman/dog-breed-identification
# Model: MobileNetV2
# Platform: Google Colab
# ============================================================


# ============================================================
# 1. INSTALL KAGGLEHUB
# ============================================================

!pip install -q kagglehub


# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import kagglehub

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report


# ============================================================
# 3. CHECK TENSORFLOW AND GPU
# ============================================================

print("TensorFlow Version:", tf.__version__)

print("GPU Available:")
print(tf.config.list_physical_devices("GPU"))


# ============================================================
# 4. DOWNLOAD DATASET DIRECTLY FROM KAGGLE
# ============================================================

path = kagglehub.dataset_download(
    "hartman/dog-breed-identification"
)

print("\nDataset downloaded successfully!")
print("Dataset path:")
print(path)


# ============================================================
# 5. DISPLAY DATASET CONTENTS
# ============================================================

print("\nDataset contents:")

for root, dirs, files in os.walk(path):

    print("\nFolder:", root)
    print("Subfolders:", dirs[:5])
    print("Files:", files[:5])

    # Stop after a few levels
    if root.count(os.sep) > path.count(os.sep) + 2:
        continue


# ============================================================
# 6. FIND LABELS.CSV
# ============================================================

labels_file = None

for root, dirs, files in os.walk(path):

    for file in files:

        if file.lower() == "labels.csv":

            labels_file = os.path.join(
                root,
                file
            )

            break

    if labels_file is not None:
        break


print("\nLabels file:")
print(labels_file)


# ============================================================
# 7. READ LABELS
# ============================================================

labels = pd.read_csv(labels_file)

print("\nFirst 5 rows:")
print(labels.head())


print("\nDataset shape:")
print(labels.shape)


print("\nNumber of breeds:")
print(labels["breed"].nunique())


# ============================================================
# 8. FIND TRAIN IMAGE FOLDER
# ============================================================

train_folder = None

for root, dirs, files in os.walk(path):

    if os.path.basename(root).lower() == "train":

        train_folder = root

        break


print("\nTrain image folder:")
print(train_folder)


# ============================================================
# 9. CREATE IMAGE PATH
# ============================================================

labels["image_path"] = labels["id"].apply(

    lambda x: os.path.join(
        train_folder,
        x + ".jpg"
    )

)


# ============================================================
# 10. CHECK WHETHER IMAGES EXIST
# ============================================================

labels["exists"] = labels["image_path"].apply(
    os.path.exists
)


print(
    "\nImages found:",
    labels["exists"].sum(),
    "/",
    len(labels)
)


# Keep only images that exist
labels = labels[
    labels["exists"] == True
].copy()


# ============================================================
# 11. DATASET INFORMATION
# ============================================================

print("\n======================================")
print("DATASET INFORMATION")
print("======================================")

print(
    "Total Images:",
    len(labels)
)

print(
    "Total Breeds:",
    labels["breed"].nunique()
)

print("\nImages per breed:")

print(
    labels["breed"].value_counts()
)


# ============================================================
# 12. TRAIN / VALIDATION SPLIT
# ============================================================

train_df, val_df = train_test_split(

    labels,

    test_size=0.20,

    random_state=42,

    stratify=labels["breed"]

)


print("\nTraining images:", len(train_df))

print("Validation images:", len(val_df))


# ============================================================
# 13. SETTINGS
# ============================================================

IMAGE_SIZE = (224, 224)

BATCH_SIZE = 32

EPOCHS = 5

FINE_TUNE_EPOCHS = 3


# ============================================================
# 14. DATA AUGMENTATION
# ============================================================

train_datagen = ImageDataGenerator(

    preprocessing_function=preprocess_input,

    rotation_range=20,

    width_shift_range=0.2,

    height_shift_range=0.2,

    zoom_range=0.2,

    horizontal_flip=True

)


# Validation data only gets preprocessing
val_datagen = ImageDataGenerator(

    preprocessing_function=preprocess_input

)


# ============================================================
# 15. CREATE TRAINING GENERATOR
# ============================================================

train_generator = train_datagen.flow_from_dataframe(

    dataframe=train_df,

    x_col="image_path",

    y_col="breed",

    target_size=IMAGE_SIZE,

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=True,

    seed=42

)


# ============================================================
# 16. CREATE VALIDATION GENERATOR
# ============================================================

validation_generator = val_datagen.flow_from_dataframe(

    dataframe=val_df,

    x_col="image_path",

    y_col="breed",

    target_size=IMAGE_SIZE,

    batch_size=BATCH_SIZE,

    class_mode="categorical",

    shuffle=False

)


# ============================================================
# 17. GET NUMBER OF CLASSES
# ============================================================

# IMPORTANT:
# DataFrameIterator does not have num_classes
# So we use class_indices

num_classes = len(
    train_generator.class_indices
)


print("\n======================================")
print("CLASS INFORMATION")
print("======================================")

print(
    "Number of dog breeds:",
    num_classes
)


# ============================================================
# 18. GET CLASS NAMES
# ============================================================

class_names = list(
    train_generator.class_indices.keys()
)


print("\nDog breeds:")

for index, breed in enumerate(class_names):

    print(
        index,
        ":",
        breed
    )


# ============================================================
# 19. LOAD PRETRAINED MOBILENETV2
# ============================================================

print("\n======================================")
print("LOADING MOBILENETV2")
print("======================================")


base_model = MobileNetV2(

    weights="imagenet",

    include_top=False,

    input_shape=(224, 224, 3)

)


print("MobileNetV2 loaded successfully!")


# ============================================================
# 20. FREEZE PRETRAINED MODEL
# ============================================================

base_model.trainable = False


# ============================================================
# 21. ADD CUSTOM CLASSIFICATION LAYERS
# ============================================================

x = base_model.output

x = GlobalAveragePooling2D()(x)

x = Dense(

    128,

    activation="relu"

)(x)

x = Dropout(

    0.5

)(x)

output = Dense(

    num_classes,

    activation="softmax"

)(x)


# ============================================================
# 22. CREATE FINAL MODEL
# ============================================================

model = Model(

    inputs=base_model.input,

    outputs=output

)


# ============================================================
# 23. COMPILE MODEL
# ============================================================

model.compile(

    optimizer=Adam(

        learning_rate=0.0001

    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)


# ============================================================
# 24. DISPLAY MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 25. TRAIN MODEL
# ============================================================

print("\n======================================")
print("STARTING TRANSFER LEARNING")
print("======================================")


history = model.fit(

    train_generator,

    validation_data=validation_generator,

    epochs=EPOCHS

)


# ============================================================
# 26. EVALUATE TRANSFER LEARNING MODEL
# ============================================================

loss, accuracy = model.evaluate(

    validation_generator

)


print("\n======================================")
print("TRANSFER LEARNING RESULTS")
print("======================================")


print(
    "Validation Loss:",
    loss
)


print(
    "Validation Accuracy: {:.2f}%".format(
        accuracy * 100
    )
)


# ============================================================
# 27. PLOT TRAINING AND VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))


plt.plot(

    history.history["accuracy"],

    label="Training Accuracy"

)


plt.plot(

    history.history["val_accuracy"],

    label="Validation Accuracy"

)


plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "Training vs Validation Accuracy"
)


plt.legend()

plt.show()


# ============================================================
# 28. PLOT TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))


plt.plot(

    history.history["loss"],

    label="Training Loss"

)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss"

)


plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "Training vs Validation Loss"
)


plt.legend()

plt.show()


# ============================================================
# 29. FINE-TUNING
# ============================================================

print("\n======================================")
print("STARTING FINE-TUNING")
print("======================================")


# Unfreeze MobileNetV2
base_model.trainable = True


# Freeze all layers except last 30
for layer in base_model.layers[:-30]:

    layer.trainable = False


# ============================================================
# 30. RECOMPILE MODEL
# ============================================================

model.compile(

    optimizer=Adam(

        learning_rate=0.00001

    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)


# ============================================================
# 31. FINE-TUNE MODEL
# ============================================================

history_fine = model.fit(

    train_generator,

    validation_data=validation_generator,

    epochs=FINE_TUNE_EPOCHS

)


# ============================================================
# 32. FINAL EVALUATION
# ============================================================

loss, accuracy = model.evaluate(

    validation_generator

)


print("\n======================================")
print("FINAL MODEL RESULTS")
print("======================================")


print(
    "Final Loss:",
    loss
)


print(
    "Final Accuracy: {:.2f}%".format(
        accuracy * 100
    )
)


# ============================================================
# 33. PLOT FINE-TUNING ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))


plt.plot(

    history_fine.history["accuracy"],

    label="Training Accuracy"

)


plt.plot(

    history_fine.history["val_accuracy"],

    label="Validation Accuracy"

)


plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "Fine-Tuning Accuracy"
)


plt.legend()

plt.show()


# ============================================================
# 34. GET PREDICTIONS
# ============================================================

validation_generator.reset()


predictions = model.predict(

    validation_generator

)


predicted_classes = np.argmax(

    predictions,

    axis=1

)


true_classes = validation_generator.classes


# ============================================================
# 35. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    true_classes,

    predicted_classes

)


plt.figure(figsize=(18, 15))


sns.heatmap(

    cm,

    annot=False,

    cmap="Blues",

    xticklabels=class_names,

    yticklabels=class_names

)


plt.xlabel(
    "Predicted Breed"
)

plt.ylabel(
    "Actual Breed"
)

plt.title(
    "Dog Breed Confusion Matrix"
)

plt.show()


# ============================================================
# 36. CLASSIFICATION REPORT
# ============================================================

print("\n======================================")
print("CLASSIFICATION REPORT")
print("======================================")


print(

    classification_report(

        true_classes,

        predicted_classes,

        target_names=class_names

    )

)


# ============================================================
# 37. SAVE MODEL
# ============================================================

model.save(

    "/content/dog_breed_mobilenetv2.h5"

)


print("\n======================================")
print("MODEL SAVED")
print("======================================")


print(
    "/content/dog_breed_mobilenetv2.h5"
)


# ============================================================
# 38. TEST A NEW IMAGE
# ============================================================

from google.colab import files

print("\n======================================")
print("UPLOAD A DOG IMAGE FOR PREDICTION")
print("======================================")


uploaded = files.upload()


# Get uploaded filename
test_image_path = list(
    uploaded.keys()
)[0]


# ============================================================
# 39. LOAD TEST IMAGE
# ============================================================

from tensorflow.keras.preprocessing import image


img = image.load_img(

    test_image_path,

    target_size=(224, 224)

)


# ============================================================
# 40. PREPROCESS IMAGE
# ============================================================

img_array = image.img_to_array(

    img

)


img_array = np.expand_dims(

    img_array,

    axis=0

)


img_array = preprocess_input(

    img_array

)


# ============================================================
# 41. PREDICT BREED
# ============================================================

prediction = model.predict(

    img_array

)


predicted_index = np.argmax(

    prediction[0]

)


predicted_breed = class_names[

    predicted_index

]


confidence = (

    prediction[0][predicted_index]

    * 100

)


# ============================================================
# 42. DISPLAY PREDICTION
# ============================================================

print("\n======================================")
print("PREDICTION RESULT")
print("======================================")


print(

    "Predicted Dog Breed:",

    predicted_breed

)


print(

    "Confidence: {:.2f}%".format(

        confidence

    )

)


# ============================================================
# 43. DISPLAY IMAGE
# ============================================================

plt.figure(figsize=(6, 6))


plt.imshow(img)


plt.axis("off")


plt.title(

    "Predicted: {}\nConfidence: {:.2f}%".format(

        predicted_breed,

        confidence

    )

)


plt.show()
```
